## 2. Dataset Overview

```python
!pip install -q kagglehub

import kagglehub

path = kagglehub.dataset_download(
    "laibaanwer/superstore-sales-dataset"
)

print("Dataset downloaded to:")
print(path)
import os

for root, dirs, files in os.walk(path):
    for file in files:
        print(os.path.join(root, file))
import pandas as pd

file_path = os.path.join(path, "SuperStoreOrders.csv")

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows and Columns:", df.shape)

display(df.head())
```

## 3. Data Quality Check

```python
# Check column names
print("Column Names:")
print(df.columns.tolist())

# Check data types and non-null values
print("\nDataset Information:")
df.info()
```

## 4. Data Cleaning

```python
# Check missing values
print("Missing Values in Each Column:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

## 5. Descriptive Statistics

```python
# Convert date columns to datetime format
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='mixed')

# Check the date range
print("Order Date Range:")
print("From:", df['order_date'].min())
print("To:", df['order_date'].max())

# Descriptive statistics
print("\nDescriptive Statistics:")
display(
    df[['sales', 'quantity', 'discount', 'profit', 'shipping_cost']].describe()
)
```

## 6. Monthly Sales Trend

```python
# Convert sales column to numeric
df['sales'] = pd.to_numeric(df['sales'], errors='coerce')

# Check the data type
print("Sales Data Type:", df['sales'].dtype)

# Check missing values created during conversion
print("Missing Sales Values:", df['sales'].isnull().sum())

import matplotlib.pyplot as plt

# Create monthly sales
monthly_sales = (
    df.groupby(df['order_date'].dt.to_period('M'))['sales']
      .sum()
      .reset_index()
)

monthly_sales['order_date'] = monthly_sales['order_date'].dt.to_timestamp()

# Plot
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales['order_date'],
    monthly_sales['sales'],
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 7. Numerical Analysis

```python
# Check the sales values that could not be converted
print("Missing/Invalid Sales Values:")
display(df[df['sales'].isnull()][['order_id', 'order_date', 'sales']].head(20))
print("Number of invalid sales records:", df['sales'].isnull().sum())
```

## 8. Categorical Analysis

```python
# Reload the original dataset without changing the values

df_raw = pd.read_csv(
    "/kaggle/input/superstore-sales-dataset/SuperStoreOrders.csv"
)

print("Original Sales Data Type:", df_raw['sales'].dtype)

# Show original sales values for records that became invalid
invalid_sales = pd.to_numeric(df_raw['sales'], errors='coerce').isnull()

print("Invalid Sales Records:", invalid_sales.sum())

display(
    df_raw.loc[invalid_sales, ['order_id', 'order_date', 'sales']].head(20)
)
```

## 9. Top 10 Products Analysis

```python
# Clean the Sales column

df['sales'] = (
    df_raw['sales']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('$', '', regex=False)
    .str.strip()
)

# Convert Sales to numeric
df['sales'] = pd.to_numeric(df['sales'], errors='coerce')

# Check result
print("Sales Data Type:", df['sales'].dtype)
print("Missing Sales Values:", df['sales'].isnull().sum())

display(df[['order_id', 'sales']].head(10))
print("Sales Descriptive Statistics:")
display(df['sales'].describe())
import matplotlib.pyplot as plt

# Create monthly sales data
monthly_sales = (
    df.groupby(df['order_date'].dt.to_period('M'))['sales']
    .sum()
)

# Convert period to string for plotting
monthly_sales.index = monthly_sales.index.astype(str)

# Plot
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()

```

## 10. Customer Segment Analysis

```python
# Check missing values in all columns

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
display(missing_values[missing_values > 0])

# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

display(missing_summary[missing_summary['Missing Values'] > 0])
# Check duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
print("Final Dataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Data Types:")
display(df.dtypes)
```

## 11. Correlation Analysis

```python
# Select numerical columns

numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns

print("Numerical Columns:")
print(list(numeric_columns))

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.hist(df['sales'], bins=50)

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))

plt.hist(df['profit'], bins=50)

plt.title('Profit Distribution')
plt.xlabel('Profit')
plt.ylabel('Frequency')

plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))

plt.hist(df['quantity'], bins=14)

plt.title('Quantity Distribution')
plt.xlabel('Quantity')
plt.ylabel('Frequency')

plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))

plt.hist(df['discount'], bins=20)

plt.title('Discount Distribution')
plt.xlabel('Discount')
plt.ylabel('Frequency')

plt.grid(True)
plt.show()
```

## 12. Profitability Analysis

```python
# Category-wise Sales

category_sales = df.groupby('category')['sales'].sum().sort_values(ascending=False)

print("Sales by Category:")
display(category_sales)

plt.figure(figsize=(8, 5))

category_sales.plot(kind='bar')

plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Segment-wise Sales

segment_sales = df.groupby('segment')['sales'].sum().sort_values(ascending=False)

print("Sales by Segment:")
display(segment_sales)

plt.figure(figsize=(8, 5))

segment_sales.plot(kind='bar')

plt.title('Sales by Segment')
plt.xlabel('Segment')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Region-wise Sales

region_sales = df.groupby('region')['sales'].sum().sort_values(ascending=False)

print("Sales by Region:")
display(region_sales)

plt.figure(figsize=(10, 5))

region_sales.plot(kind='bar')

plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Ship Mode distribution

ship_mode_count = df['ship_mode'].value_counts()

print("Orders by Ship Mode:")
display(ship_mode_count)

plt.figure(figsize=(8, 5))

ship_mode_count.plot(kind='bar')

plt.title('Orders by Ship Mode')
plt.xlabel('Ship Mode')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

## 13. Business Insights

```python
# Top 10 Products by Sales

top_products = (
    df.groupby('product_name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print("Top 10 Products by Sales:")
display(top_products)

plt.figure(figsize=(12, 6))

top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product')

plt.grid(axis='x')
plt.tight_layout()
plt.show()

# Top 10 Customers by Sales

top_customers = (
    df.groupby('customer_name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print("Top 10 Customers by Sales:")
display(top_customers)

plt.figure(figsize=(12, 6))

top_customers.sort_values().plot(kind='barh')

plt.title('Top 10 Customers by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Customer')

plt.grid(axis='x')
plt.tight_layout()
plt.show()

```

## 14. Business Recommendations

```python
# Revenue by Product Category

category_revenue = (
    df.groupby('category')['sales']
      .sum()
      .sort_values(ascending=False)
)

print("Revenue by Product Category:")
display(category_revenue)

plt.figure(figsize=(8, 5))

category_revenue.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Revenue by Sub-Category

subcategory_revenue = (
    df.groupby('sub_category')['sales']
      .sum()
      .sort_values(ascending=False)
)

print("Revenue by Sub-Category:")
display(subcategory_revenue)



```

```python
# Customer segment distribution

segment_count = df['segment'].value_counts()

print("Number of Customers by Segment:")
display(segment_count)

plt.figure(figsize=(8, 5))

segment_count.plot(kind='bar')

plt.title('Customer Segment Distribution')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Records')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

# Sales by customer segment

segment_sales = (
    df.groupby('segment')['sales']
      .sum()
      .sort_values(ascending=False)
)

print("Sales by Customer Segment:")
display(segment_sales)

plt.figure(figsize=(8, 5))

segment_sales.plot(kind='bar')

plt.title('Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
correlation_data = df[
    ['sales', 'quantity', 'discount', 'profit', 'shipping_cost']
]

# Calculate correlation
correlation_matrix = correlation_data.corr()

print("Correlation Matrix:")
display(correlation_matrix)

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()

# Profit by Category

category_profit = (
    df.groupby('category')['profit']
      .sum()
      .sort_values(ascending=False)
)

print("Profit by Category:")
display(category_profit)

plt.figure(figsize=(8, 5))

category_profit.plot(kind='bar')

plt.title('Profit by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)

plt.grid(axis='y')
plt.tight_layout()
plt.show()

```

## 15. Conclusion

```python
# Find key business insights from the dataset

top_category = df.groupby('category')['sales'].sum().idxmax()
top_category_sales = df.groupby('category')['sales'].sum().max()

top_product = df.groupby('product_name')['sales'].sum().idxmax()
top_product_sales = df.groupby('product_name')['sales'].sum().max()

top_region = df.groupby('region')['sales'].sum().idxmax()
top_region_sales = df.groupby('region')['sales'].sum().max()

top_profit_category = df.groupby('category')['profit'].sum().idxmax()
top_profit = df.groupby('category')['profit'].sum().max()

print("Top Sales Category:", top_category)
print("Top Category Sales:", round(top_category_sales, 2))

print("\nTop Product:", top_product)
print("Top Product Sales:", round(top_product_sales, 2))

print("\nTop Sales Region:", top_region)
print("Top Region Sales:", round(top_region_sales, 2))

print("\nMost Profitable Category:", top_profit_category)
print("Category Profit:", round(top_profit, 2))


# Overall business KPIs

total_sales = df['sales'].sum()
total_profit = df['profit'].sum()
total_quantity = df['quantity'].sum()

print("===== BUSINESS KPIs =====")
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Profit Margin (%):", round((total_profit / total_sales) * 100, 2))
```

```python
# Create quarterly sales data

quarterly_sales = (
    df.groupby(df['order_date'].dt.to_period('Q'))['sales']
      .sum()
      .reset_index()
)

# Convert quarter period to string
quarterly_sales['order_date'] = quarterly_sales['order_date'].astype(str)

print("Quarterly Sales:")
display(quarterly_sales)

plt.figure(figsize=(12, 6))

plt.plot(
    quarterly_sales['order_date'],
    quarterly_sales['sales'],
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')

plt.grid(True)
plt.tight_layout()
plt.show()
```
